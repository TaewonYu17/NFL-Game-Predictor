```python
import time
import datetime
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
from io import StringIO

# List of NFL team abbreviations as used in pro-football-reference URLs
teams = [
    'Manchester-City', 'Arsenal', 'Liverpool', 'Chelsea', 'Tottenham', 'Brentford', 'Brighton', 'Aston-Villa', 'Fulham', 'Crystal-Palace', 'West-Ham', 'Wolves', 'Bournemouth', 'Everton'
]
current_year = datetime.datetime.now().year
years = list(range(current_year - 6, current_year))  # Last 5 years

all_games = []

# Set up Selenium WebDriver
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

for team in teams:
    for year in years:
        url = f"https://fbref.com/en/squads/b8fd03ef/{year - 1}-{year}/matchlogs/c9/schedule/{team}-Scores-and-Fixtures-Premier-League"
        print(f"Scraping {url}")
        driver.get(url)
        time.sleep(5)

        soup = BeautifulSoup(driver.page_source, 'html.parser')
        table = soup.find('table', id='matchlogs_for')
       
        if table is None:
            print(f"No table found for {team} {year}")
            continue

        df = pd.read_html(StringIO(str(table)))[0]
        df['Team'] = team
        df['Year'] = year
        all_games.append(df)

driver.quit()

# Combine all dataframes
if all_games:
    full_df = pd.concat(all_games, ignore_index=True)
    full_df = full_df[~(full_df == full_df.columns).all(axis=1)]
    full_df.to_csv('epl_schedule_results_last_5_years.csv', index=False)
    print("Saved to epl_schedule_results_last_5_years.csv")
else:
    print("No data scraped.")
```
